```python
nlp = spacy.load("es_core_news_sm")
```

<p><b>2.</b> Limpieza, tokenización y reducción del vocabulario.</p>

```python
comentarios = pd.read_csv("comentarios.csv")
comentarios.head()
```

```python
comentarios.info()
```

```python
comentarios.shape
```

```python
stopword = set(stopwords.words("spanish"))
```

```python
patron = RegexpTokenizer(r"(?![.,;])[A-Za-zÁÉÍÓÚáéíóúÑñ]+|:\(|:\)|[\U0001F300-\U0001FAFF]|[❤❣]")
excepciones = {"no", "muy", "mal", "bien", "ok", "si"}
```

```python
def limpieza_tokenizacion(texto):
    texto = str(texto).lower()
    tokens = patron.tokenize(texto)
    tokens_limpios = []
    for txt in tokens:
        if (txt in excepciones) or (txt not in stopword):
            tokens_limpios.append(txt)
    return tokens_limpios
```

```python
comentarios["tokens"] = comentarios["comentario"].apply(limpieza_tokenizacion)
comentarios[["comentario", "tokens"]].head()
```

```python
def lematizacion(tokens):
    texto = " ".join(tokens)
    doc = nlp(texto)
    lemma = []
    for token in doc:
        if token.text in [":)", ":("] or not token.is_space:
            if not token.is_space:
                lemma.append(token.lemma_.lower())
    return lemma
```

```python
comentarios["lemmas"] = comentarios["tokens"].apply(lematizacion)
comentarios[["tokens", "lemmas"]].head()
```

```python
stemmer = SnowballStemmer("spanish")

def stemming(tokens):
    return [stemmer.stem(txt)
    for txt in tokens]
```

```python
comentarios["stems"] = comentarios["tokens"].apply(stemming)
comentarios[["tokens", "lemmas", "stems"]].head()
```

```python
comentarios.to_pkl("comentarios_procesados.pkl", index=False, encoding="utf-8")
```
