```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
comentarios = pd.read_csv("comentarios_procesados.csv")
```

```python
comentarios.head()
```

```python
comentarios.info
```

```python
comentarios[["texto_lemmas", "tipo"]].head()
```

```python
x = comentarios["texto_lemmas"]
y = comentarios["tipo"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)
```

```python
vectorizador = TfidfVectorizer()
```

```python
x_train_tfidf = vectorizador.fit_transform(x_train)
x_test_tfidf = vectorizador.transform(x_test)
```

```python
print(f"(cantidad de comentarios, términos únicos) train: {x_train_tfidf.shape}")
print(f"(cantidad de comentarios, términos únicos) test: {x_test_tfidf.shape}")
```

```python
vocabulario = vectorizador.get_feature_names_out()
print(vocabulario)
```

```python
def mostrar_tfidf(orden):
    # Esta función permitirá ingresar un índice de comentario, y obtendrá por respuesta
    # el peso tfidf de cada una de sus palabras que le componen
    fila = comentarios[comentarios["orden"] == orden]
    if fila.empty:
        print("Comentario no encontrado, por favor, ingrese otro valor")
        return
    texto = fila["texto_lemmas"].iloc[0]
    categoria = fila["tipo"].iloc[0]
    vector = vectorizador.transform([texto])
    tfidf = pd.DataFrame({
        "Palabra": vectorizador.get_feature_names_out(),
        "Peso TF-IDF": vector.toarray().flatten()
    })
    tfidf = (
        tfidf[tfidf["Peso TF-IDF"] > 0]
        .sort_values(by="Peso TF-IDF", ascending=False)
        .reset_index(drop=True)
    )
    print("=" * 80)
    print(f"Comentario N° {orden}")
    print(f"Categoría: {categoria}")
    print("-" * 80)
    print(texto)
    print("=" * 80)
    return tfidf
```

```python
"""Ingresar un valor del 1 al 600 (cantidad de comentarios) para obtener un rápido análisis del mismo"""
mostrar_tfidf(95)
```
